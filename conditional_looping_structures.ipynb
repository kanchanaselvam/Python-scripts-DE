```python
#read the given list of file, first check whether the file is there in the directory
import pandas as pd
import os
dirpath="C:\\Users\\DELL\\Documents\\bfsi_dataset\\"
all_file_list=os.listdir(dirpath)
intended_file_list=['accounts.csv', 'branches.csv', 'loans.csv']
print(f"all_file_list: {all_file_list}")
print(f"intended_file_list: {intended_file_list}")
for i in all_file_list:
    print(f"******** reading {i} file *******")
    if i in intended_file_list:
        print(f" {i} file present in {intended_file_list}")
        df=pd.read_csv(dirpath+i)
        print(df.head(3))
    else:
        print(f" {i} file  not present in {intended_file_list}")
else:
    print("reached the end")
```

```python
#using continue
list_of_states=['TN','PO','KA','KL','AP','TL','GO','MP','RJ','DH']
car_price=400000
tax_pc=18
for i in list_of_states:
    if i in ('PO','GO'):
        print(f"car price is {car_price}, as no tax is applied for state {i}")
        continue
        car_total_amount=car_price+car_price*tax_pc/100
        print(f"car price is {car_total_amount}, with tax is applied, as state is {i}")
```

```python
#using break
# when password is incorrect give attempt for 3 times
# when password is correct , come out of the loop
password='kanchana123'
i=1
while(i<=3):
    userpwd=input("enter your password: ")
    if userpwd==password:
        print("you are allowed to access the application")
        break
    else:
        print(f"password is not match, try again,{3-i} attempts remaining")
    i=i+1
    print(i)
else:
    print(f"maximum {i-1} attempts reached")
```

```python
# using exception in the while loop, password should be integer
try:
    password='123456'
    i=1
    while(i<=3):
        userpwd=int(input("enter your password: "))
        if userpwd==password:
            print("you are allowed to access the application")
            break
        else:
            print(f"password is not match, try again,{3-i} attempts remaining")
        i=i+1
        print(i)
    else:
        print(f"maximum {i-1} attempts reached")
except Exception as e:
        print(f"exception raise is ",e)
```
